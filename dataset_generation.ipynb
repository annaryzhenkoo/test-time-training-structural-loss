```python
size_dst = 20000
```

```python
digits_number = 4
```

```python
import random

x = random.randint(0, 999)
y = random.randint(0, 999)

print(x)
print(y)
print(x + y)
print(f"{x}+{y}=")
```

```python
dst = []

for i in range(size_dst):

    if digits_number == 3:
        a = random.randint(0, 999)
        b = random.randint(0, 999)

    if digits_number == 4:
        a = random.randint(1000, 9999)
        b = random.randint(1000, 9999)

    addition = f"{a}+{b}={a+b}"

    dst.append(addition)
```

```python
import pandas as pd

df = pd.DataFrame(dst)
df
```

```python
df.drop_duplicates(inplace=True)
len(df)
```

```python
df.to_csv(f"data_{digits_number}_{len(df)}.csv", index=False)
```
